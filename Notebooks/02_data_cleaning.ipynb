```python
import pandas as pd
import numpy as np

# Load original raw dataset
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# 1. Missing / Blank Values Handling in TotalCharges
df['TotalCharges'] = df['TotalCharges'].replace(" ", np.nan)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
df['TotalCharges'].fillna(df['TotalCharges'].median(), inplace=True)

# 2. Duplicate CustomerIDs Check
df.drop_duplicates(subset=['customerID'], inplace=True)

# 3. Data Types Cast
df['tenure'] = df['tenure'].astype(int)
df['SeniorCitizen'] = df['SeniorCitizen'].astype(int)

# 4. Save Clean Dataset (Original raw CSV overwrite nahi karni hai)
df.to_csv("../data/telco_customer_churn_clean.csv", index=False)
print("Clean dataset successfully saved to data/telco_customer_churn_clean.csv")
```
